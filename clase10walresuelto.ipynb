```python
# ### 8. Reto integrador: pedido con entrega 
# **Consigna:** incorpora “Retiro en local” y “Entrega a domicilio” mediante dos `Radiobutton` que compartan un `StringVar`.
# 1. Añade un campo de dirección.
# 2. Exige dirección únicamente para entrega a domicilio.
# 3. Agrega G. 10.000 de envío cuando se elija domicilio.
# 4. Calcula el descuento sobre los productos y después suma el envío.
# 5. Muestra modalidad, dirección cuando corresponda, subtotal, descuento, envío y total final.
# 6. Restablece “Retiro en local” y dirección vacía al limpiar.
# **Prueba:** productos por G. 45.000, descuento del 10 % y domicilio → G. 50.500. Sin descuento y con retiro → G. 45.000.
import tkinter as tk
from tkinter import messagebox


def guaranies(valor):
    return f"G. {valor:,}".replace(",", ".")


def calcular():
    try:
        precio = int(entrada_precio.get().strip())
        cantidad = int(entrada_cantidad.get().strip())

        if precio <= 0 or cantidad <= 0:
            raise ValueError("Precio y cantidad deben ser mayores que cero.")

        direccion = entrada_direccion.get().strip()
        domicilio = modalidad.get() == "Entrega a domicilio"

        if domicilio and not direccion:
            raise ValueError("Ingresa la dirección para la entrega.")

        subtotal = precio * cantidad
        descuento = subtotal // 10 if aplicar_descuento.get() else 0
        envio = 10000 if domicilio else 0
        total = subtotal - descuento + envio

        texto = f"Modalidad: {modalidad.get()}\n"
        if domicilio:
            texto += f"Dirección: {direccion}\n"

        texto += (
            f"Subtotal: {guaranies(subtotal)}\n"
            f"Descuento: {guaranies(descuento)}\n"
            f"Envío: {guaranies(envio)}\n"
            f"Total final: {guaranies(total)}"
        )
        resultado.config(text=texto)

    except ValueError as error:
        resultado.config(text="")
        messagebox.showerror(
            "Datos incorrectos",
            f"Usa precio y cantidad como enteros positivos.\n\n"
            f"Detalle: {error}"
        )


def limpiar():
    if not messagebox.askyesno(
        "Confirmar limpieza", "¿Quieres limpiar el formulario?"
    ):
        return

    for entrada in (entrada_producto, entrada_precio, entrada_cantidad, entrada_direccion,):
        entrada.delete(0, tk.END)

    aplicar_descuento.set(False)
    modalidad.set("Retiro en local")
    resultado.config(text="")
    entrada_producto.focus_set()


ventana = tk.Tk()
ventana.title("Pedido con entrega")

seccion = tk.LabelFrame(ventana, text="Pedido", padx=15, pady=15)
seccion.pack(padx=20, pady=15, fill="x")

tk.Label(seccion, text="Nombre:").grid(row=0, column=0, sticky="w")
entrada_producto = tk.Entry(seccion, width=30)
entrada_producto.grid(row=0, column=1, pady=5)

tk.Label(seccion, text="Precio (G.):").grid(row=1, column=0, sticky="w")
entrada_precio = tk.Entry(seccion, width=30)
entrada_precio.grid(row=1, column=1, pady=5)

tk.Label(seccion, text="Cantidad:").grid(row=2, column=0, sticky="w")
entrada_cantidad = tk.Entry(seccion, width=30)
entrada_cantidad.grid(row=2, column=1, pady=5)

aplicar_descuento = tk.BooleanVar(value=False)
tk.Checkbutton(
    seccion,
    text="Aplicar descuento del 10 %",
    variable=aplicar_descuento,
).grid(row=3, column=0, columnspan=2, sticky="w", pady=10)

modalidad = tk.StringVar(value="Retiro en local")

tk.Radiobutton(
    seccion,
    text="Retiro en local",
    variable=modalidad,
    value="Retiro en local",
).grid(row=4, column=0, columnspan=2, sticky="w")

tk.Radiobutton(
    seccion,
    text="Entrega a domicilio",
    variable=modalidad,
    value="Entrega a domicilio",
).grid(row=5, column=0, columnspan=2, sticky="w")

tk.Label(seccion, text="Dirección:").grid(row=6, column=0, sticky="w")
entrada_direccion = tk.Entry(seccion, width=30)
entrada_direccion.grid(row=6, column=1, pady=5)

botones = tk.Frame(ventana)
botones.pack(pady=5)

tk.Button(
    botones, text="Calcular", command=calcular, width=12
).pack(side="left", padx=5)

tk.Button(
    botones, text="Limpiar", command=limpiar, width=12
).pack(side="left", padx=5)

resultado = tk.Label(ventana, text="", justify="left", font=("Arial", 12))
resultado.pack(padx=20, pady=15)

entrada_producto.focus_set()
ventana.mainloop()
```
